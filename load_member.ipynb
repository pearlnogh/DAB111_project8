### Data Preprocessing: Create Database and load CSV into SQLite

```python
import pandas as pd
import sqlite3

# Read CSV
dataset = pd.read_csv('Gym Members Tracking.csv')

# Create SQLite DB and gym_members_tracking table
connection = sqlite3.connect('gym_members_tracking.db')

dataset.to_sql('gym_members_tracking', connection, if_exists='replace', index=False)


cursor = connection.cursor()
cursor.execute('SELECT * FROM gym_members_tracking LIMIT 5')
for i in cursor.fetchall():
    print(i)


connection.close()
```
